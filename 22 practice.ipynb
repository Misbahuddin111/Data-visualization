```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
batsman = pd.read_csv("Datasets/sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],label='sharma')
plt.legend()
plt.xlabel('session')
plt.ylabel('runs')
plt.title("sharma and kholi runs")
```

```python
# labels title # color , linewidth ,linestyle
plt.plot(batsman['index'],batsman['V Kohli'],label='virat',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],label='sharma',linewidth=2)
plt.legend()
plt.xlabel('session')
plt.ylabel('runs')
plt.title("sharma and kholi runs")
```

```python
# marker and marksize
plt.plot(batsman['index'],batsman['V Kohli'],label='virat',linestyle='solid',marker='o')
plt.plot(batsman['index'],batsman['RG Sharma'],label='sharma',linewidth=2,marker='d')
plt.legend()
plt.xlabel('session')
plt.ylabel('runs')
plt.title("sharma and kholi runs")
plt.grid()
plt.show()
```
