```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',size='size')
```

```python
sns.scatterplot(data=tips,x=tips['total_bill'],y=tips['tip'],hue='day')
plt.title("toatl bill and tip graph")
plt.show()
```

```python

sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],row=tips['sex'],col=tips['smoker'])
```

```python
sns.relplot(data=tips,x=tips['total_bill'],y=tips['tip'],row=tips['sex'],col=tips['smoker'],kind='line')
plt.show()
```

### cat plot

```python
sns.catplot(data=tips,x=tips['day'],y=tips['total_bill'],hue=tips['sex'],jitter=False)
```

(The categorical plots do not currently support size or style semantics). Each different categorical plotting function handles the hue semantic differently. For the scatter plots, it is only necessary to change the color of the points:

```python
# swarm plot
sns.catplot(data=tips,x=tips['day'],y=tips['total_bill'],hue=tips['sex'],kind='swarm')
```

```python
sns.catplot(data=tips,x=tips['smoker'],y=tips['total_bill'],order=['No', 'Yes'],hue='sex')
plt.show()
```
